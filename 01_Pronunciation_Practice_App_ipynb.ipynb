```python
# ============================================================
# PRONUNCIATION PRACTICE APP
# Google Colab - Upload Audio Version
# ============================================================

!pip -q install SpeechRecognition rapidfuzz

import speech_recognition as sr
from rapidfuzz.fuzz import ratio
from google.colab import files
from IPython.display import display, HTML


# ------------------------------------------------------------
# TEXT CLEANING
# ------------------------------------------------------------

def clean_text(text):
    text = text.lower()

    for symbol in ".,!?;:'\"":
        text = text.replace(symbol, "")

    return " ".join(text.split())


# ------------------------------------------------------------
# SPEECH TO TEXT
# ------------------------------------------------------------

def speech_to_text(audio_file):

    recognizer = sr.Recognizer()

    with sr.AudioFile(audio_file) as source:
        audio = recognizer.record(source)

    try:
        return recognizer.recognize_google(audio)

    except sr.UnknownValueError:
        return ""

    except sr.RequestError:
        return ""


# ------------------------------------------------------------
# PRONUNCIATION ANALYSIS
# ------------------------------------------------------------

def analyze_pronunciation(expected, spoken):

    expected = clean_text(expected)
    spoken = clean_text(spoken)

    if not spoken:
        return 0, [], expected.split()

    score = ratio(expected, spoken)

    expected_words = expected.split()
    spoken_words = spoken.split()

    correct = []
    incorrect = []

    for word in expected_words:

        if word in spoken_words:
            correct.append(word)
        else:
            incorrect.append(word)

    return score, correct, incorrect


# ------------------------------------------------------------
# FEEDBACK
# ------------------------------------------------------------

def feedback(score):

    if score >= 90:
        return "Excellent pronunciation! Keep it up."

    elif score >= 75:
        return "Good pronunciation. Practice once more for better accuracy."

    elif score >= 50:
        return "Fair pronunciation. Try speaking slowly and clearly."

    else:
        return "Needs improvement. Listen carefully and try again."


# ------------------------------------------------------------
# APPLICATION HEADER
# ------------------------------------------------------------

display(HTML("""
<div style="
background:#f5f7fb;
padding:30px;
border-radius:20px;
text-align:center;
font-family:Arial;
">

<h1>Pronunciation Practice App</h1>

<p>
Text & Speech Analysis - Application 1
</p>

<p>
Practice your pronunciation and receive an accuracy score.
</p>

</div>
"""))


# ------------------------------------------------------------
# GET TARGET SENTENCE
# ------------------------------------------------------------

target = input("\nEnter the sentence you want to practice: ")

print("\nTarget sentence:")
print(target)


# ------------------------------------------------------------
# UPLOAD AUDIO
# ------------------------------------------------------------

print("\nNow record yourself saying the sentence.")

print("Save the recording as a WAV audio file.")

print("\nClick the button below to upload your recording.")

uploaded = files.upload()


# ------------------------------------------------------------
# GET UPLOADED FILE
# ------------------------------------------------------------

audio_file = list(uploaded.keys())[0]

print("\nAudio uploaded successfully:")
print(audio_file)


# ------------------------------------------------------------
# SPEECH RECOGNITION
# ------------------------------------------------------------

print("\nAnalyzing your speech...")

spoken = speech_to_text(audio_file)


# ------------------------------------------------------------
# DISPLAY SPEECH
# ------------------------------------------------------------

print("\n" + "=" * 60)

print("PRONUNCIATION ANALYSIS")

print("=" * 60)

print("\nExpected sentence:")
print(target)

print("\nRecognized speech:")

if spoken:
    print(spoken)
else:
    print("Speech could not be recognized.")


# ------------------------------------------------------------
# ANALYSIS
# ------------------------------------------------------------

score, correct_words, incorrect_words = analyze_pronunciation(
    target,
    spoken
)


# ------------------------------------------------------------
# RESULT
# ------------------------------------------------------------

print("\n" + "=" * 60)

print("RESULT")

print("=" * 60)

print(f"\nPronunciation Accuracy: {score:.2f}%")


print("\nCorrect words:")

if correct_words:
    print(", ".join(correct_words))
else:
    print("None")


print("\nWords to practice:")

if incorrect_words:
    print(", ".join(incorrect_words))
else:
    print("None")


print("\nFeedback:")

print(feedback(score))


# ------------------------------------------------------------
# VISUAL RESULT
# ------------------------------------------------------------

display(HTML(f"""

<div style="
max-width:700px;
margin:30px auto;
padding:30px;
border-radius:20px;
background:white;
border:1px solid #ddd;
font-family:Arial;
">

<h2 style="text-align:center;">
Pronunciation Result
</h2>

<hr>

<p>
<b>Expected:</b><br>
{target}
</p>

<p>
<b>You said:</b><br>
{spoken if spoken else "Speech not recognized"}
</p>

<div style="
text-align:center;
font-size:42px;
font-weight:bold;
margin:25px;
">

{score:.2f}%

</div>

<p style="text-align:center;font-size:18px;">
{feedback(score)}
</p>

</div>

"""))
```
